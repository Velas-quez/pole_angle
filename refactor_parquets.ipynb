```python
import polars as pl
import os
```

```python
folder_paths = ["data/train/preprocessed_mix", "data/train/preprocessed_babs_new"]
```

```python
os.listdir(folder_paths[1])
```

```python
dfs = {
    f.removesuffix(".parquet"): pl.concat(
        [pl.read_parquet(os.path.join(folder_paths[0], f)),
        pl.read_parquet(os.path.join(folder_paths[1], f))],
        how="vertical"
    ) for f in os.listdir(folder_paths[0])}
```

```python
for k, df in dfs.items():
    dfs[k] = df.with_columns(pl.col("bag_name").str.strip_prefix("pole_balancing_data_").str.strip_suffix(".bag").str.to_datetime("%Y-%m-%d-%H-%M-%S", strict=False).alias("date")).with_columns(
        pl.when(pl.col("date") < pl.date(2026,3,1)).then(pl.lit("velas")).otherwise(pl.lit("babs")).alias("user")
    ).drop("date")
```

```python
dfs["evtac"]
```

```python
for k, df in dfs.items():
    fpath = f"data/train/preprocessed_mix_updated/{k}.parquet"
    print(f"saving {fpath}")
    df.write_parquet(fpath)
```
